```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(20, 18))

# ---------- KPI VALUES ----------
total_sales = df['Sales'].sum()
total_orders = df['Order_ID'].nunique()
total_customers = df['Customer_ID'].nunique()
total_products = df['Product_ID'].nunique()

# ---------- KPI CARDS ----------
fig.text(0.15, 0.95, f"Total Sales\n${total_sales:,.2f}",
         ha='center', va='center', fontsize=16, fontweight='bold', color='#2c3e50')

fig.text(0.38, 0.95, f"Total Orders\n{total_orders:,}",
         ha='center', va='center', fontsize=16, fontweight='bold', color='#2c3e50')

fig.text(0.62, 0.95, f"Total Customers\n{total_customers:,}",
         ha='center', va='center', fontsize=16, fontweight='bold', color='#2c3e50')

fig.text(0.85, 0.95, f"Total Products\n{total_products:,}",
         ha='center', va='center', fontsize=16, fontweight='bold', color='#2c3e50')


# ---------- 1. CATEGORY SALES ----------
ax1 = fig.add_axes([0.08, 0.52, 0.38, 0.32])

ax1.bar(category_sales['Category'], category_sales['Sales'], color='#3498db', edgecolor='black', alpha=0.8)
ax1.set_title('Category-wise Sales', fontsize=14, fontweight='bold', pad=10)
ax1.set_xlabel('Category', fontsize=12)
ax1.set_ylabel('Sales ($)', fontsize=12)
ax1.grid(axis='y', linestyle='--', alpha=0.5)


# ---------- 2. REGION SALES ----------
ax2 = fig.add_axes([0.55, 0.52, 0.38, 0.32])

region_sales = df.groupby('Region')['Sales'].sum()
ax2.bar(region_sales.index, region_sales.values, color='#2ecc71', edgecolor='black', alpha=0.8)
ax2.set_title('Region-wise Sales', fontsize=14, fontweight='bold', pad=10)
ax2.set_xlabel('Region', fontsize=12)
ax2.set_ylabel('Sales ($)', fontsize=12)
ax2.grid(axis='y', linestyle='--', alpha=0.5)


# ---------- 3. MONTHLY SALES ----------
ax3 = fig.add_axes([0.08, 0.10, 0.38, 0.32])

ax3.plot(month_sales.index, month_sales.values, marker='o', color='#e74c3c', linewidth=2, markersize=6)
ax3.set_title('Monthly Sales Trend', fontsize=14, fontweight='bold', pad=10)
ax3.set_xlabel('Month', fontsize=12)
ax3.set_ylabel('Sales ($)', fontsize=12)
ax3.tick_params(axis='x', rotation=45)
ax3.grid(True, linestyle='--', alpha=0.5)


# ---------- 4. TOP 10 PRODUCTS (WITH TRUNCATED NAMES TO PREVENT OVERLAP) ----------
ax4 = fig.add_axes([0.55, 0.10, 0.38, 0.32])

top_10_products = (
    df.groupby('Product_Name')['Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

# Truncate product names to keep the Y-axis clean and avoid overlaps
truncated_names = [name[:30] + '...' if len(name) > 30 else name for name in top_10_products.index]

ax4.barh(
    truncated_names,
    top_10_products.values,
    color='#9b59b6',
    edgecolor='black',
    alpha=0.8
)

ax4.set_title('Top 10 Products by Sales', fontsize=14, fontweight='bold', pad=10)
ax4.set_xlabel('Sales ($)', fontsize=8)
ax4.grid(axis='x', linestyle='--', alpha=0.5)
ax4.invert_yaxis()  # Keep highest sales at the top

plt.show()
```
